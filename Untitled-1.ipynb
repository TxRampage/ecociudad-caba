```python
import os
import shutil
import random
from pathlib import Path
from PIL import Image

# 1. Definición de clases según el modelo de datos de EcoCiudad
CLASSES = ['cardboard', 'glass', 'metal', 'paper', 'plastic', 'trash']
CLASS_MAP = {name: idx for idx, name in enumerate(CLASSES)}

RAW_DIR = Path("data/raw/trashnet")
OUTPUT_DIR = Path("data/dataset_yolo")
SPLIT_RATIO = 0.8  # 80% train, 20% val

def setup_yolo_directories():
    for split in ['train', 'val']:
        (OUTPUT_DIR / 'images' / split).mkdir(parents=True, exist_ok=True)
        (OUTPUT_DIR / 'labels' / split).mkdir(parents=True, exist_ok=True)

def process_trashnet():
    setup_yolo_directories()
    
    for category in CLASSES:
        cat_dir = RAW_DIR / category
        if not cat_dir.exists():
            continue
            
        images = [f for f in cat_dir.iterdir() if f.suffix.lower() in ['.jpg', '.jpeg', '.png']]
        random.seed(42)
        random.shuffle(images)
        
        split_idx = int(len(images) * SPLIT_RATIO)
        train_imgs = images[:split_idx]
        val_imgs = images[split_idx:]
        
        for split, img_list in [('train', train_imgs), ('val', val_imgs)]:
            for img_path in img_list:
                # Copiar imagen
                dest_img = OUTPUT_DIR / 'images' / split / img_path.name
                shutil.copy(img_path, dest_img)
                
                # Generar etiqueta YOLO (caja centrada para objeto principal)
                # Formato: <class_id> <x_center> <y_center> <width> <height>
                class_id = CLASS_MAP[category]
                label_path = OUTPUT_DIR / 'labels' / split / f"{img_path.stem}.txt"
                
                with open(label_path, 'w') as f:
                    f.write(f"{class_id} 0.5 0.5 0.9 0.9\n")

    # Crear data.yaml
    yaml_content = f"""path: {OUTPUT_DIR.resolve()}
train: images/train
val: images/val

names:
"""
    for idx, name in enumerate(CLASSES):
        yaml_content += f"  {idx}: {name}\n"
        
    with open(OUTPUT_DIR / "data.yaml", "w") as f:
        f.write(yaml_content)
        
    print("Dataset YOLO generado exitosamente en:", OUTPUT_DIR)

if __name__ == "__main__":
    process_trashnet()
```
